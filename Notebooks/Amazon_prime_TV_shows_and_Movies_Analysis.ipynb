 ## Amazon Prime TV Shows and Movies Analysis — Project Summary

This project focuses on performing an in-depth Exploratory Data Analysis (EDA) on the Amazon Prime Movies and TV Shows dataset to uncover meaningful insights related to content distribution, audience ratings, genre trends, regional production, popularity patterns, and platform growth over time. The primary objective of this analysis was to transform raw and unstructured entertainment data into an analysis-ready dataset and build an interactive business-oriented Tableau dashboard capable of supporting strategic decision-making.

The project began with a comprehensive data cleaning and preprocessing pipeline using Python libraries such as Pandas, NumPy, Matplotlib, and Seaborn. Two separate datasets containing title information and credits/cast information were initially inspected for missing values, duplicates, inconsistent data types, and structural issues. Duplicate records were identified and removed to improve data quality. Missing values in important columns such as IMDb scores, TMDB popularity, votes, seasons, and age certifications were handled using statistical techniques like median imputation and custom labels such as “Not Rated” and “No Character Info.”

Several feature engineering techniques were applied to enhance analytical capabilities. New columns such as content_age, decade, runtime_category, rating_category, genre_category, country_region, imdb_vote_category, and tmdb_popu_category were created to simplify advanced analysis and storytelling. Complex list-like columns including genres and production countries were converted into usable Python lists, enabling better categorization and segmentation of the content library.

To make the dataset suitable for Tableau visualization, the credits dataset was aggregated and summarized to avoid duplicate title rows caused by one-to-many relationships between movies/shows and cast members. A final merged dataset containing content metadata, engineered features, and summarized cast information was created and exported as clean CSV files.

The Tableau dashboard was designed to provide both high-level KPIs and detailed analytical insights. Important KPIs included Total Titles, Average IMDb Score, Average TMDB Popularity, Total Movies, and Total TV Shows. Multiple interactive visualizations were developed to answer key business questions regarding Amazon Prime’s content strategy and audience engagement.

Genre analysis revealed that categories related to Entertainment & Fun and Emotional & Family content dominate the platform’s library, indicating strong audience preference toward mainstream and emotionally engaging entertainment. Runtime analysis showed that medium-length content tends to maintain more stable IMDb ratings compared to extremely short or very long content. Rating distribution analysis demonstrated that most titles fall within the “Good” and “Very Good” rating categories, suggesting an overall balanced content quality across the platform.

A dual-axis visualization comparing content volume and average IMDb score across release years highlighted the relationship between platform growth and content quality. The analysis indicated that Amazon Prime significantly expanded its content library in recent years while maintaining relatively stable audience ratings. Regional and map-based visualizations further demonstrated that North America and Asia contribute a major share of the platform’s content production, while certain European regions exhibit comparatively stronger average ratings.

Scatter plot analysis between TMDB popularity and IMDb scores revealed that highly popular content does not always guarantee better audience ratings, emphasizing the importance of balancing commercial appeal with content quality. Additional insights related to recent versus older content suggested that newer titles dominate the platform’s growth strategy, reflecting evolving streaming consumption patterns.

Overall, this project demonstrates strong capabilities in data wrangling, feature engineering, exploratory data analysis, data storytelling, and dashboard development. The final solution successfully converts raw entertainment data into actionable business insights and provides a scalable analytical framework for OTT platform content analysis.

````python
# Write your code to make your dataset analysis ready.
''' after getting the info of two datasets,a nd checking wheather they
 have nulls or duplicates or not, which they have, now I will directly start
 wranglling the data as I have specified the necesary steps including importing
 libraries, checking info, duplicates, and missing or null earlier'''

# --------------------------------------------------------------------------
# ----------------REMOVING DIPLICATES BEFORE DATA WRANGLING PIPELINE------------

titles_dup_before = df.duplicated().sum()
credits_dup_before = df_2.duplicated().sum()

df.drop_duplicates(inplace=True)
df_2.drop_duplicates(inplace=True)

# Duplicates removed after handleing nulls and duplicates
print("Duplicates removed from titles:", titles_dup_before)
print("Duplicates removed from credits:", credits_dup_before)



#-------- DATA WRANGLLING PIPELINE----------
# HANDLEING MISSING VALUES - TITLES AKA DF
#-----------------------------------------

# filling age_certification with Unknown
df['age_certification'] = df['age_certification'].fillna('Not Rated')

# filling in seasons null value with 0
df['seasons'] = df['seasons'].fillna(0)

# filling null values of imdb_score with median()
df['imdb_score'] = df['imdb_score'].fillna(df['imdb_score'].median())

# filling in missing votes with median()
df['imdb_votes'] = df['imdb_votes'].fillna(df['imdb_votes'].median())

# handleing tmdb_popularity with median()
df['tmdb_popularity'] = df['tmdb_popularity'].fillna(df['tmdb_popularity'].median())

# filling in missing values of tmdb_score with median()
df['tmdb_score'] = df['tmdb_score'].fillna(df['tmdb_score'].median())

# dropping rows where title is missing to avoid unnecesary missing value without tittle
df.dropna(subset=['title'],inplace = True)



#------------------------------------------------------------------------
# HANDLEING MISSING OR NULL VALUES IN CREDITS AKA DF_2
#----------------------------------------------------------------------

# filling in missing charector names with not_applicable
df_2['character'] = df_2['character'].fillna('No Character Info')

# dropping rows where name is missing to avoid uncecesary values without name not needed
df_2.dropna(subset=['name'], inplace =True)


#------------------------------------------------------------------------------------------
''' converting columns to proper numeric types so differet type of analysis can be performed
on the numeric columns'''

# fixing dtypes of numeric cols
num_cols = ['seasons', 'imdb_score',
           'imdb_votes', 'tmdb_score',
           'tmdb_popularity', 'release_year','runtime' ]

for col in num_cols:
  df[col] = pd.to_numeric(df[col], errors = 'coerce')


# --------------------------------------------------------------------------------
# -------------FEATURE ENGINEERING------------------------------------
# ----------------------------------------------------------------------

# creating new column called content age to understand the age of the content
df['content_age'] = 2026 - df['release_year']

# creating decade age to identify from which decade the content belongs
df['decade'] = (df['release_year'] // 10)*10


# creating runtime category to see how much runtime a content has
df['runtime_category'] = pd.cut(
    df['runtime'],
    bins=[0,60,120,180,500],
    labels=['Short', 'Medium', 'Long', 'Very Long']
)

# creating rating category for rating content
df['rating_category'] = pd.cut(
    df['imdb_score'],
    bins=[0,5,7,8.5,10],
    labels=['Bad','Good','Very Good', 'Excellent']
)

# creating a popularity category using custom functions
def popularity_bucket(popularity):
  if popularity < 10:
    return 'Low'
  elif popularity <= 40:
    return 'Medium'
  elif popularity <= 70:
    return 'High'
  else :
    return 'Very High'


df['tmdb_popu_category'] = df['tmdb_popularity'].apply(popularity_bucket)

# creating imdb votes category using custom function
def vote_category(votes):
  if votes < 1000:
    return 'Low votes'
  elif votes <= 10000:
    return 'Medium votes'
  elif votes <= 85000:
    return 'High votes'
  else :
    return 'Very High votes'


df['imdb_vote_category'] = df['imdb_votes'].apply(vote_category)


#---------------------------------------------------------------------
# CONVERTING STR LISTS TO REAL LISTS
#---------------------------------------------------------------------
import ast

def safe_convert(x):
    if isinstance(x, list):
        return x
    try:
        return ast.literal_eval(x) if pd.notnull(x) else []
    except:
        return []

df['genres'] = df['genres'].apply(safe_convert)
df['production_countries'] = df['production_countries'].apply(safe_convert)

# creating genre category directly without explode
def genre_category(genres):
    genres = [str(g).lower() for g in genres]

    if any(g in ['drama', 'romance', 'family'] for g in genres):
        return 'Emotional & Family'

    elif any(g in ['comedy', 'music', 'reality'] for g in genres):
        return 'Entertainment & Fun'

    elif any(g in ['thriller', 'crime', 'horror', 'mystery'] for g in genres):
        return 'Suspense & Crime'

    elif any(g in ['action', 'war', 'western', 'fantasy', 'scifi'] for g in genres):
        return 'Action & Adventure'

    elif any(g in ['documentation', 'history'] for g in genres):
        return 'Educational'

    else:
        return 'Other'


df['genre_category'] = df['genres'].apply(genre_category)


# creating country region directly without explode
def country_region(countries):
    countries = [str(c).upper() for c in countries]

    if any(c in ['US', 'CA', 'MX'] for c in countries):
        return 'North America'

    elif any(c in ['IN', 'CN', 'JP', 'KR'] for c in countries):
        return 'Asia'

    elif any(c in ['GB', 'FR', 'DE', 'IT', 'ES'] for c in countries):
        return 'Europe'

    elif any(c in ['BR', 'AR', 'CL'] for c in countries):
        return 'South America'

    elif any(c in ['AU', 'NZ'] for c in countries):
        return 'Oceania'

    else:
        return 'Other'


df['country_region'] = df['production_countries'].apply(country_region)


#--------------------------------------------------------------------------------
# -------------------- CREATING COLUMNS AFTER CONVERSON-------------------------

#creating final two columns called number of genre and is recent
df['num_genres'] = df['genres'].apply(len)
df['is_recent'] = np.where(df['release_year'] >= 2018,1,0)


#---------------------------------------------------------------------------------
# ---------------RESETTING INDEX --------------------------------------
#--------------------------------------------------------------------------------------

df.reset_index(drop=True, inplace=True)
df_2.reset_index(drop=True, inplace=True)




#-------------------------------------------------------------------------
# -------------- NOW MERGING THE TITLES AKA DF AND CREDITS AKA DF_2--------------
#----------------------------------------------------------------------------

# aggregate credits
credits_summary = df_2.groupby('id').agg({
    'name': lambda x: ', '.join(x.astype(str).unique()[:5]),
    'role': lambda x: ', '.join(x.astype(str).unique()),
    'character': lambda x: ', '.join(x.astype(str).unique()[:5])
}).reset_index()

# merging two dataframes
merged_df = pd.merge(df, credits_summary, on='id', how='left')

# converting list columns to strings for saving/tableau
merged_df['genres'] = merged_df['genres'].astype(str)
merged_df['production_countries'] = merged_df['production_countries'].astype(str)

print("Merged dataset shape:", merged_df.shape)
print("Duplicates:", merged_df.duplicated().sum())




#-----------------------------------------------------------------------------
#---------------- POST CLEANNING AFTER MERGEING THE TWO DATASETS--------------
#---------------------------------------------------------------------------------

# filling in missing roles after merge with unknown
merged_df['role'] = merged_df['role'].fillna('No Role Info')

# filling in merged_df character with not applicable
merged_df['character'] = merged_df['character'].fillna('No Character Info')

# filling in missing persons name with unknown
merged_df['name'] = merged_df['name'].fillna('No Cast Info')


#-----------------------------------------------------------------------------------
# ------------------FINAL DUPLICATES CHECH AFTER POST CLEANNING-----------------------
#-----------------------------------------------------------------------------------

# checking DUPLICATES AFTER POST CLEANNING THE MERGE DATA
print("\n Duplicated after merge :",
      merged_df.drop(columns=['genres', 'production_countries']).duplicated().sum())


#---------------------------------------------------------------------------------------
#---------------- FINAL CHECK AFTER POST CLEANNING MERGE DATA-------------------------
#------------------------------------------------------------------------------------------

print("\n final Tittles Dataset Shape :", df.shape)
print("\n final Credits Dataset Shape :", df_2.shape)
print("\n final Merged Dataset Shape :", merged_df.shape)

# --------------------------------------------------------------------------------------------
#-----The final structural information of the datasets after cleaning-----------
df.info()
df_2.info()
merged_df.info()


# ============================================================
# SAVE CLEANED DATASETS TO GOOGLE DRIVE
# Run this AFTER complete data wrangling is finished
# ============================================================

print("\nData Wrangling Completed Successfully")

#  Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Create folder if it doesn't exist
import os

save_path = '/content/drive/MyDrive/My_Projects/'
os.makedirs(save_path, exist_ok=True)

print("Folder ready:", save_path)

#  Save cleaned datasets
df.to_csv(save_path + 'cleaned_titles.csv', index=False)
df_2.to_csv(save_path + 'cleaned_credits.csv', index=False)
merged_df.to_csv(save_path + 'cleaned_merged_dataset.csv', index=False)

print("Files saved successfully in Google Drive")

#  Verify files
print("Saved files:")
print(os.listdir(save_path))
````
